```python
from statsmodels.stats.power import NormalIndPower
from statsmodels.stats.proportion import proportion_effectsize
# Параметри тесту
baseline = 0.037 # поточна конверсія 3,7%
mde = 0.20 # MDE: +20% відносних (3,7% → 4,44%)
alpha = 0.05  # рівень значущості
power = 0.80 # статистична потужність
# Цільова конверсія тестової групи
target = baseline * (1 + mde)  # 0.037 * 1.20 = 0.0444
# Розмір ефекту (effect size) для пропорцій
effect_size = proportion_effectsize(baseline, target)
print(f"Effect size: {effect_size:.4f}")
# Розрахунок розміру вибірки на одну групу
analysis = NormalIndPower()
n_per_group = analysis.solve_power(
   effect_size=effect_size,
   alpha=alpha,
   power=power,
  alternative='two-sided'
)
n_per_group = int(n_per_group) + 1
print(f"Розмір вибірки на групу: {n_per_group}")
print(f"Загальний розмір вибірки: {n_per_group * 2}")
```
